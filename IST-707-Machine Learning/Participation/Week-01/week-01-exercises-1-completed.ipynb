1. Read the `2019-yellow-taxi-trips.csv` data set into Pandas. Create a `pairs plot` visualization of 5-6 non-categorical, non-date columns of interest.

```python
# https://ist707.s3.us-east-2.amazonaws.com/data/2019-yellow-taxi-trips.csv
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# load
df = pd.read_csv("../data/2019-yellow-taxi-trips.csv")
df.head()
```

```python
df.info()
```

```python
# Selecting numerical columns of interest
columns_of_interest = [
    "trip_distance",
    "fare_amount",
    "tip_amount",
    "tolls_amount",
    "total_amount",
    "congestion_surcharge"
]

# Sampling a smaller subset to improve performance in plotting
sample_df = df.dropna()[columns_of_interest].sample(n=1000, random_state=1)

# Plotting
sns.pairplot(sample_df)
plt.suptitle("Pair Plot of NYC Yellow Taxi Trip Metrics (Sample of 1000)", y=1.02)
plt.show()
```
