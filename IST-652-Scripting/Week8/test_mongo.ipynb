```python
from pymongo import MongoClient
import pprint
import bson

client = MongoClient('localhost', 27017)

# show existing databases
print(client.list_database_names())
```

```python
# get the bbcoll collection
bball = client.bball
bbcoll = bball.bbcoll
```

```python
one_tweet = bbcoll.find_one({"id_str": "846463866863276032"})
```

```python
pprint.pprint(one_tweet)
```

```python
for tweet in bbcoll.find().limit(5):
    pprint.pprint(tweet)
```

```python
regx = bson.regex.Regex('iPhone</a>$')
iphone_tweets = bbcoll.find({'source': regx })
for tweet in iphone_tweets.sort("country_code").limit(5):
    pprint.pprint(tweet)
```

```python
client.close()
```
